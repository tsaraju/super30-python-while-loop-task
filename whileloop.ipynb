```python
'''
Calculate the sum of digits of a number.
Example
Input: 5832
Output: 18
'''
n = int(input("Enter a number: "))
sum = 0
while n != 0:
    r = n % 10
    sum += r
    n = n // 10

print("The sum of digits of the number is:", sum)
```

```python
'''
Reverse an integer using a while loop.
Example
Input: 12345
Output: 54321
'''
n = int(input("Enter a number: "))
rev = ''
while n != 0:
    r = n % 10
    rev = rev + str(r)
    n = n // 10
    
print("The reverse of the number is:", rev)
```

```python

#Count the number of digits in an integer.
#Input - 55
n = int(input("Enter a number: "))
cnt = 0
while n != 0:
    cnt += 1
    n = n // 10
print("The number of digits in the number is:", cnt)
```

```python
#Calculate factorial using while.
#Input 3
n = int(input("Enter a number: "))
fact = 1

while n > 1:
    fact = fact * n
    n = n - 1
print("The factorial of the number is:", fact)
```

```python
'''
Build a menu-driven calculator
1  Add
2  Subtract
3  Multiply
4  Divide
5  Exit

The menu should continue until the user chooses Exit.
'''
while True:
    opt = int(input("Enter an option in number: "))
    if opt == 5:
        break
    elif opt == 1:
        print("Add option selected")
    elif opt == 2:
        print("Subtract option selected")
    elif opt == 3:
        print("Multiply option selected")
    elif opt == 4:
        print("Divide option selected")
    else:
        print("Invalid option selected")
```

```python
'''
Create an ATM menu using while where the application continues 
running until the user explicitly chooses Exit
'''
while True:
    opt = input("Enter an option: ")
    if opt == 'Exit':
        break
    elif opt == 'Deposit':
        print("Money is being deposited")
    elif opt == 'Withdraw':
        print("Money is being withdrawn")
    elif opt == 'Balance':
        print("Balance is being displayed")
    elif opt == 'MiniStatement':
        print("Mini statement is being printed")    
    elif opt == 'ChangePin':
        print("Pin is being changed")
    elif opt == 'ResetPin':
        print("Pin is being reset")
    elif opt == 'Transfer':
        print("Money is being transferred")
    else:
        print("Invalid option selected")
```
